```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.integrate import cumulative_trapezoid
from sklearn.gaussian_process import GaussianProcessRegressor
from sklearn.gaussian_process.kernels import RBF, WhiteKernel
import emcee
import corner

# Constants
c = 299792.458  # km/s
H0 = 67.66  # Planck value

# Loading the  Pantheon SN dataset
pantheon = pd.read_csv('lcparam_full_long.txt', delim_whitespace=True, comment='#',
                       names=["name","zcmb","zhel","dz","mb","dmb","x1","dx1","color","dcolor",
                              "3rdvar","d3rdvar","cov_m_s","cov_m_c","cov_s_c","set","ra","dec","biascor"])
z_sn = pantheon['zcmb'].values.astype(float)
mu_sn = pantheon['mb'].values.astype(float)
mu_err = pantheon['dmb'].values.astype(float)

#Loading full covariance matrix data
data = np.loadtxt("sys_full_long.txt")
data = data[:1048*1048]
C_sys = data.reshape((1048, 1048))
C_stat = np.diag(mu_err**2)
C_full = C_stat + C_sys
Cinv = np.linalg.inv(C_full)

# Loading H(z) data
hz_data = pd.read_csv('Hz_data.csv')
z_hz = hz_data['z'].values
Hz = hz_data['Hz'].values
Hz_err = hz_data['err'].values

# Gaussian Process Regression for  to reconstruct both the distance modulus μ(z) and the Hubble parameter H(z)
xgrid = np.linspace(0.01, 1.85, 200).reshape(-1,1)
z_grid = xgrid.flatten()

kernel_mu = 1.0 * RBF(0.3) + WhiteKernel(0.1)
gp_mu = GaussianProcessRegressor(kernel=kernel_mu, alpha=mu_err**2)
gp_mu.fit(z_sn.reshape(-1,1), mu_sn)
mu_pred, _ = gp_mu.predict(xgrid, return_std=True)

dL_fid = 10**((mu_pred - 25)/5)  # Mpc

kernel_hz = 1.0 * RBF(0.3) + WhiteKernel(1.0)
gp_hz = GaussianProcessRegressor(kernel=kernel_hz, alpha=Hz_err**2)
gp_hz.fit(z_hz.reshape(-1,1), Hz)
Hz_pred, Hz_std = gp_hz.predict(xgrid, return_std=True)
Hz_pred = Hz_pred.flatten()

integral = cumulative_trapezoid(1.0 / Hz_pred, z_grid, initial=0)
dC = c * integral
dA_flat = dC / (1 + z_grid)

# defining function for dA for non-flat universe
def dA_nonflat(z_vals, Hz_vals, Omega_k):
    integral = cumulative_trapezoid(1.0 / Hz_vals, z_vals, initial=0)
    dC = c * integral
    DH = c / H0
    sqrtOk = np.sqrt(np.abs(Omega_k))
    if Omega_k > 0:
        Sk = np.sinh(sqrtOk * dC / DH) / sqrtOk
    elif Omega_k < 0:
        Sk = np.sin(sqrtOk * dC / DH) / sqrtOk
    else:
        Sk = dC / DH
    return DH * Sk / (1 + z_vals)

# defining function for different  Models with definitions
def eta_model(z, params, model_type):
    if model_type == 'P1':
        return params[0] * z
    elif model_type == 'P2':
        return params[0] * z + params[1] * z**2
    elif model_type == 'P3':
        return params[0] * (z / (1 + z)) + params[1]
    elif model_type == 'P4':
        return params[0] * np.log(1 + z) + params[1] * (np.log(1 + z))**2
    else:
        raise ValueError("Unknown eta model")

def MB_model(z, params, model_type):
    if model_type == 'M1':
        return params[0]
    elif model_type == 'M2':
        return params[0] + params[1] * z
    elif model_type == 'M3':
        return params[0] + params[1] * (z / (1 + z))
    elif model_type == 'M4':
        return params[0] + params[1] * np.log(1 + z)
    else:
        raise ValueError("Unknown MB model")

#  defing Log-likelihoods function for both eta and MB models
def log_likelihood_eta(params, z, mu_obs, model_type, is_flat):
    if is_flat:
        eta_params = params[:-1]
        MB = params[-1]
        dA = np.interp(z, z_grid, dA_flat)
    else:
        eta_params = params[:-2]
        MB = params[-2]
        Omega_k = params[-1]
        dA = np.interp(z, z_grid, dA_nonflat(z_grid, Hz_pred, Omega_k))
    eta_vals = eta_model(z, eta_params, model_type)
    if np.any(eta_vals <= 0) or np.any(dA <= 0):
        return -np.inf
    mu_model = 5 * np.log10((dA * (1 + z)**2) / eta_vals) + 25 + MB
    delta = (mu_obs - mu_model).reshape(-1)
    return -0.5 * delta @ Cinv @ delta

def log_likelihood_MB(params, z, mu_obs, model_type, is_flat):
    if is_flat:
        MB_vals = MB_model(z, params, model_type)
        dA = np.interp(z, z_grid, dA_flat)
    else:
        Omega_k = params[-1]
        MB_params = params[:-1]
        MB_vals = MB_model(z, MB_params, model_type)
        dA = np.interp(z, z_grid, dA_nonflat(z_grid, Hz_pred, Omega_k))
    if np.any(dA <= 0):
        return -np.inf
    mu_model = 5 * np.log10(dA * (1 + z)**2) + 25 + MB_vals
    delta = (mu_obs - mu_model).reshape(-1)
    return -0.5 * delta @ Cinv @ delta

# defining function for priors for the eta and mb
def log_prior_eta(params, model_type, is_flat):
    bounds_dict = {
        'P1': [(-3, 3)],
        'P2': [(-3, 3), (-2, 2)],
        'P3': [(-3, 3), (-3, 3)],
        'P4': [(-3, 3), (-2, 2)]
    }
    bounds = bounds_dict[model_type] + [(-21, -17)]
    if not is_flat:
        bounds.append((-2.5, 2.5))
    for val, (low, high) in zip(params, bounds):
        if not (low < val < high):
            return -np.inf
    return 0.0

def log_prior_MB(params, model_type, is_flat):
    bounds_dict = {
        'M1': [(-21, -17)],
        'M2': [(-21, -17), (-2, 2)],
        'M3': [(-21, -17), (-2, 2)],
        'M4': [(-21, -17), (-2, 2)]
    }
    bounds = bounds_dict[model_type]
    if not is_flat:
        bounds.append((-2.5, 2.5))
    for val, (low, high) in zip(params, bounds):
        if not (low < val < high):
            return -np.inf
    return 0.0

# defing Posteriors function for eta and mb
def log_prob_eta(params, z, mu_obs, model_type, is_flat):
    lp = log_prior_eta(params, model_type, is_flat)
    if not np.isfinite(lp):
        return -np.inf
    return lp + log_likelihood_eta(params, z, mu_obs, model_type, is_flat)

def log_prob_MB(params, z, mu_obs, model_type, is_flat):
    lp = log_prior_MB(params, model_type, is_flat)
    if not np.isfinite(lp):
        return -np.inf
    return lp + log_likelihood_MB(params, z, mu_obs, model_type, is_flat)

# defing mcmc runner function   for both eta and mb models
def run_eta_model(model_type='P1', is_flat=True, nsteps=10000, walkers=100):
    n_eta = {'P1': 1, 'P2': 2, 'P3': 2, 'P4': 2}[model_type]
    ndim = n_eta + 1 if is_flat else n_eta + 2
    init = np.random.normal(loc=[0.0]*ndim, scale=[0.1]*ndim, size=(walkers, ndim))
    if is_flat:
        init[:,-1] += -19.3
    else:
        init[:,-2] += -19.3
    sampler = emcee.EnsembleSampler(walkers, ndim, log_prob_eta,
                                     args=(z_sn, mu_sn, model_type, is_flat))
    print(f"Running MCMC for model {model_type} (eta), flat={is_flat}.")
    sampler.run_mcmc(init, nsteps, progress=True)
    samples = sampler.get_chain(discard=nsteps//5, flat=True)
    labels = ['eta0', 'eta1'] if n_eta == 2 else ['eta0']
    labels += ['MB'] if is_flat else ['MB', 'Omega_k']
    corner.corner(samples, labels=labels, show_titles=True)
    plt.show()
    for i, label in enumerate(labels):
        print(f"{label} = {np.median(samples[:,i]):.4f} ± {np.std(samples[:,i]):.4f}")
    return samples
def run_MB_model(model_type='M1', is_flat=True, nsteps=10000, walkers=100):
    n_MB = {'M1': 1, 'M2': 2, 'M3': 2, 'M4': 2}[model_type]
    ndim = n_MB if is_flat else n_MB + 1
    init = np.random.normal(loc=[-19.3] + [0.0]*(ndim - 1), scale=[0.1]*ndim, size=(walkers, ndim))
    sampler = emcee.EnsembleSampler(walkers, ndim, log_prob_MB,
                                     args=(z_sn, mu_sn, model_type, is_flat))
    print(f"Running MCMC for model {model_type} (MB), flat={is_flat}.")
    sampler.run_mcmc(init, nsteps, progress=True)
    samples = sampler.get_chain(discard=nsteps//5, flat=True)
    labels = ['MB0'] if model_type == 'M1' else ['MB0', 'MB1']
    if not is_flat:
        labels.append('Omega_k')
    corner.corner(samples, labels=labels, show_titles=True)
    plt.show()
    for i, label in enumerate(labels):
        print(f"{label} = {np.median(samples[:,i]):.4f} ± {np.std(samples[:,i]):.4f}")
    return samples

#Running all the  eta and MB models
for model in ['P1', 'P2', 'P3', 'P4']:
    run_eta_model(model, is_flat=True)
    run_eta_model(model, is_flat=False)

for model in ['M1', 'M2', 'M3', 'M4']:
    run_MB_model(model, is_flat=True)
    run_MB_model(model, is_flat=False)
```
